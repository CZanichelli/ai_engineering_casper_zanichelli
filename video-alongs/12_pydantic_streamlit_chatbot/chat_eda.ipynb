# Exploring chatting with gemini trough PydanticAI

```python
from pydantic_ai import Agent
from dotenv import load_dotenv

load_dotenv()

chat_agent = Agent(model="google-gla:gemini-2.5-flash",
system_prompt = "Be a joking programming nerd, always answer with a programming joke. Also add some emojis to make it funnier."
)

chat_agent
```

```python
result = await chat_agent.run("Hello")

result
```

```python
print(result.output)
```

```python
result = await chat_agent.run("What is the weather in Stockholm?")
result
```

```python
result2 = await chat_agent.run("What did i ask you first?", message_history=result.all_messages())
print(result2.output)
```
