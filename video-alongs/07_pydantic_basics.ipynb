```python
from pydantic import BaseModel

class PokemonListResponse(BaseModel):
    count: int
    results: list[dict]

pokemons = PokemonListResponse.model_validate(data)

pokemons
```

```python
type(pokemons)
```

```python
pokemons.count
```

```python
pokemons.results
```

```python
data = requests.get("https://pokeapi.co/api/v2/pokemon-species/pikachu").json()

data.keys()
```

```python
for key, values in data.items():
    print(f"{key:<30} {type(values)}")
```

```python
class Pokemon(BaseModel):
    id: int
    name: str
    base_happiness: int
    capture_rate: int
    evolution_chain: dict
    gender_rate: int
    generation: dict

pikachu = Pokemon.model_validate(data)

pikachu
```

```python
pikachu.evolution_chain
```

```python
pikachu.generation
```

```python
serialized_pikacu = pikachu.model_dump_json()

serialized_pikacu
```

```python
type(serialized_pikacu)
```

```python
print(serialized_pikacu)
```

```python
print(pikachu.model_dump_json(indent =2))
```
