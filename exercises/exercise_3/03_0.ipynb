b) Create a BaseModel for a Person with the fields name, age, email, favourite pet. Add appropriate validation in each fields. Tips: you can use built-in EmailStr type in pydantic for validating email. Try out your Person class by instantiating it with different types of values for the fields to see proper validations.

```python
from pydantic import EmailStr, BaseModel

class Person(BaseModel):
    name: str
    age: int
    email: EmailStr
    favourite_pet: str
```

c) Use normal python class to replicate what you have created in b), i.e. create a Person class with proper input validation.

```python
#Normal python class
import re
class Person:
    def __init__(self, name, age, email, favourite_pet):
        
        #Name validation
        if not isinstance(name, str) or not name.strip():
            raise ValueError("Name must be a non type string")
        self.name = name
        
        #Age validation
        if not isinstance(age, int) or age <0:
            raise ValueError("Age must be a positive integer")

        self.age = age
        
        #Email validation
        
        email_regex = r"[^@]+@[^@]+\.[^@]+"
        if not isinstance(email, str) or not re.match(email_regex, email):
            raise ValueError("Invalid email format")
        self.email = email

        #Favourite pet validation
        if not isinstance(favourite_pet, str) or not favourite_pet.strip():
            raise ValueError("Favourite pet must be a non-empty string")
        self.favourite_pet = favourite_pet
```

```python
#Valid person with normal python class
person1 = Person(name="Casper", age=27, email="CasperZanichelli98@gmail.com", favourite_pet="Dog")

print(person1.__dict__)
```

```python
#invalid person with normal python class

try:
    person2 = Person("Casper", -5, "not-an-email", "")
except ValueError as err:
    print(err)
```
